## Test løsningen din

Simuleringer gir litt ulike svar hver gang, så testen godtar svar som ligger nær nok den riktige verdien.

```python
import math

def test_monte_carlo():
    feil = 0
    pi = anslag_pi(200_000)
    if pi is None or abs(pi - math.pi) > 0.02:
        print(f"❌ anslag_pi(200000) ga {pi!r}, forventet ca. {math.pi:.4f}")
        feil += 1
    for personer, fasit in [(10, 0.117), (23, 0.507), (50, 0.970)]:
        svar = felles_bursdag(personer, 20_000)
        if svar is None or abs(svar - fasit) > 0.02:
            print(f"❌ felles_bursdag({personer}, 20000) ga {svar!r}, forventet ca. {fasit}")
            feil += 1
    for bytte, fasit in [(True, 2/3), (False, 1/3)]:
        svar = monty_hall(bytte, 20_000)
        if svar is None or abs(svar - fasit) > 0.02:
            print(f"❌ monty_hall({bytte}, 20000) ga {svar!r}, forventet ca. {fasit:.3f}")
            feil += 1
    if feil == 0:
        print("✅ Alle testene bestått!")

test_monte_carlo()
```
